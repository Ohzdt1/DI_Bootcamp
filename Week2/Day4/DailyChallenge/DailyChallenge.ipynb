```python
from collections import Counter
import string
import re
```

```python
class Text:
    def __init__(self, text):
        self.text = text
    
    def word_frequency(self, word):
        words = self.text.split()
        word_count = Counter(words)
        return word_count.get(word, "Word not found in text")
    
    def most_common_word(self):
        words = self.text.split()
        word_count = Counter(words)
        return word_count.most_common(1)[0][0] if word_count else None
    
    def unique_words(self):
        words = self.text.split()
        return list(set(words))
    
    @classmethod
    def from_file(cls, filename):
        with open(filename, 'r', encoding='utf-8') as file:
            text = file.read()
        return cls(text)
    
class TextModification(Text):
    def remove_punctuation(self):
        return self.text.translate(str.maketrans('', '', string.punctuation))
    
    def remove_stop_words(self):
        stop_words = set("a an the and or but if then because as on at to in with of for is was were be been being has have had do does did not so nor yet can will shall should would could may might must".split())
        words = self.text.split()
        filtered_words = [word for word in words if word.lower() not in stop_words]
        return " ".join(filtered_words)
    
    def remove_special_characters(self):
        return re.sub(r'[^A-Za-z0-9\s]', '', self.text)

```

```python
text_sample = "A good book would sometimes cost as much as a good house."
text_analysis = Text(text_sample)

print("Word Frequency of 'good':", text_analysis.word_frequency("good"))
print("Most Common Word:", text_analysis.most_common_word())
print("Unique Words:", text_analysis.unique_words())
```

```python
txt_analysis = Text.from_file("the_stranger.txt")
print("Most Common Word in File:", txt_analysis.most_common_word())
```

```python
text_mod = TextModification("A good book would sometimes cost as much as a good house. I am testing Ponctuation and Special characters!")
print("Text without punctuation:", text_mod.remove_punctuation())
print("Text without stop words:", text_mod.remove_stop_words())
print("Text without special characters:", text_mod.remove_special_characters())
```
