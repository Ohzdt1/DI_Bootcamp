*In an artificial neural network (ANN), what is the function of an activation function?*

They act like a switch, turning the perceptron’s output into something useful. Activation functions determine whether a neuron should be “activated” (answer: yes) or not (answer : no) based on the weighted sum of inputs

```python
#🌟 Exercise 2 : Building a Simple Perceptron Decision System

# Weights and bias
def perceptron_decision(temperature, rain):
    weight_temp = 0.6
    weight_rain = 0.4
    bias = 2

    # Calculate weighted sum
    weighted_sum = (temperature * weight_temp) + (rain * weight_rain) + bias

    # Step activation function
    if weighted_sum > 20:
        return 1
    else:
        return 0

# Case 1: Temperature = 70°F, Rain = 0 (No)
output1 = perceptron_decision(70, 0)
print("Case 1 Output (70°F, No Rain):", "Go Outside" if output1 == 1 else "Stay Inside")

# Case 2: Temperature = 50°F, Rain = 1 (Yes)
output2 = perceptron_decision(50, 1)
print("Case 2 Output (50°F, Rainy):", "Go Outside" if output2 == 1 else "Stay Inside")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
```

```python
#🌟 Exercise 3 : Building a Simple Neural Network with TensorFlow/Keras

# Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# Normalizing teh data
x_train = x_train / 255.0
x_test = x_test / 255.0

# One-hot encode 
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
# Build the model
model = Sequential([
    Flatten(input_shape=(28, 28)),         # Input layer
    Dense(128, activation='relu'),         # Hidden layer
    Dense(10, activation='softmax')        # Output layer
])

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train, epochs=10, batch_size=32, validation_split=0.1)

# Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {accuracy:.4f}")
```

```python
#🌟 Exercise 4 : Forward Propagation Calculation

#  Inputs
x1 = 2000  # square footage
x2 = 3     # number of bedrooms

# Weights
w1 = 0.5
w2 = 0.7

# Bias
b = 50000

# Step 1: Calculate the weighted sum (z)
z = (x1 * w1) + (x2 * w2) + b

# Step 2: Apply ReLU activation
def relu(x):
    return max(0, x)

# Final prediction
predicted_price = relu(z)

# Print result
print("Weighted sum (z):", z)
print("Predicted House Price after ReLU:", predicted_price)
```

```python
#🌟 Exercise 5 : Implementing Forward and Backward Propagation in Python

# Initialize input data (features)
x = np.array([4, 80])  # 4 hours studied, previous test score: 80

# Initialize weights and bias
w = np.array([0.6, 0.3])  # Initial weights
b = 10  # Initial bias

# Forward Propagation
def forward_propagation(x, w, b):
    z = np.dot(x, w) + b  # Weighted sum
    return z  # Linear activation (No ReLU here, it's a regression task)

# Compute prediction
y_pred = forward_propagation(x, w, b)
y_true = 85  # Actual exam score

# Compute Loss (Mean Squared Error)
loss = 0.5 * (y_true - y_pred) ** 2

# Compute Gradients
grad_w = -(y_true - y_pred) * x  # Partial derivatives with respect to weights
grad_b = -(y_true - y_pred)  # Partial derivative with respect to bias

# Update Weights and Bias
learning_rate = 0.01
w_new = w - learning_rate * grad_w
b_new = b - learning_rate * grad_b

# Print Results
print("Initial Prediction:", y_pred)
print("Loss:", loss)
print("Updated Weights:", w_new)
print("Updated Bias:", b_new)
```

**Explain why updating weights using gradient descent reduces the error.**

It does this by changing the weights in the opposite direction of the gradient this in turn causes the predictions to descend towards minimizing the error and leading to more accurate predictions. In other words it looks at what was wrong with the predictions and slowly changes the parameters to make the predictions more accurate in turn minimizing the error. 

```python
# Modified input and parameters
x = np.array([4, 80])
y_true = 85

w = np.array([0.2, 0.1])  # Modified weights
b = 10
learning_rate = 0.05      # Increased learning rate

# Forward Propagation
def forward_propagation(x, w, b):
    return np.dot(x, w) + b

y_pred = forward_propagation(x, w, b)

# Loss
loss = 0.5 * (y_true - y_pred) ** 2

# Gradients
grad_w = -(y_true - y_pred) * x
grad_b = -(y_true - y_pred)

# Update weights and bias
w_new = w - learning_rate * grad_w
b_new = b - learning_rate * grad_b

# Output
print("Initial Prediction:", y_pred)
print("Loss:", loss)
print("Updated Weights:", w_new)
print("Updated Bias:", b_new)
```

```python
#🌟 Exercise 6 : Visualizing Predictions on the MNIST Dataset
 
# 1. Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 2. Normalize the data (scale pixel values to 0–1)
x_train = x_train / 255.0
x_test = x_test / 255.0

# 3. One-hot encode the labels
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)

# 4. Build the model
model = Sequential([
    Flatten(input_shape=(28, 28)),      # Flatten 28x28 images
    Dense(128, activation='relu'),      # Hidden layer
    Dense(10, activation='softmax')     # Output layer (10 classes)
])

# 5. Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 6. Train the model
model.fit(x_train, y_train_cat, epochs=5, batch_size=32, validation_split=0.1)

# 7. Make predictions
predictions = model.predict(x_test)
predicted_labels = np.argmax(predictions, axis=1)

# 8. Visualize some predictions
def visualize_predictions(images, true_labels, predicted_labels, num=10):
    plt.figure(figsize=(15, 4))
    for i in range(num):
        plt.subplot(1, num, i+1)
        plt.imshow(images[i], cmap='gray')
        plt.title(f"True: {true_labels[i]}\nPred: {predicted_labels[i]}")
        plt.axis('off')
    plt.tight_layout()
    plt.show()

visualize_predictions(x_test, y_test, predicted_labels)
```
